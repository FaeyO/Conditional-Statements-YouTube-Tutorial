```python
age = 18

if age >= 18:
    print("You are eligible to vote!")
```

```python
score = 75

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
```

```python
if condition:
    # Code for if
else:
    # Code for else
```

```python
temperature = 15

if temperature > 30:
    print("It's hot!")
else:
    print("It's cool!")
```

```python
username = "admin"
password = "1235"

if username == "admin" or password == "1234":
    print("Access granted!")
else:
    print("Access denied!")
```

```python
is_admin = True

if not is_admin:
    print("Access denied. Admins only!")
else:
    print("Access granted. Welcome, Admin!")
    
```

```python
not True = False
```

```python
age = 17
citizen = True

if age >= 18:
    if citizen:
        print("You are eligible to vote!")
    else:
        print("You need to be a citizen to vote.")
else:
    print("You are too young to vote.")
```

```python
number = int(input("Enter a number: "))

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    print("The number is zero.")
```
